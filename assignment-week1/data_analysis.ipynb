# Shopping Dataset — Data Exploration & Cleaning
**Dataset:** Shopping Dataset 
**Source:** [Kaggle – anvitkumar/shopping-dataset](https://www.kaggle.com/datasets/anvitkumar/shopping-dataset)  
**Goal:** Load raw data, explore it, fix quality issues, derive new columns, and export a clean version into new dataset.

```python
df.drop_duplicates(inplace=True)
```

```python
df['size'] = df['size'].fillna('Not Applicable')
```

```python
df['price'] = df.groupby('product_name')['price'].transform(lambda x: x.fillna(x.median()))
```

```python
df['quantity'].fillna(df['quantity'].median(), inplace=True)
```

```python
for col in ['color', 'payment_method', 'customer_gender', 'location']:
    df[col].fillna(df[col].mode()[0], inplace=True)
```

```python
df['customer_age'].fillna(df['customer_age'].median(), inplace=True)
df['rating'].fillna(df['rating'].median(), inplace=True)
```

```python
df['purchase_date'] = pd.to_datetime(df['purchase_date'])
```

```python
print("selecting single column 'price':")
print(df['price'])
```

```python
print( "selecting multiple columns:")
print(df[['product_name', 'price', 'quantity']])
      
```

```python
print("selecting all the numerical columns:")
df.select_dtypes(include=['number'])
```
